#Outliers


```python
#before outlier handling
num_cols = ['Age', 'Current_semester', 'CGPA', 'Cost_without_study']

plt.figure(figsize=(12, 8))

for i, col in enumerate(num_cols, 1):
    plt.subplot(3, 3, i)
    plt.boxplot(df_student[col])
    plt.title(col.replace("_", " ").title())

plt.tight_layout()
plt.show()
```

```python
num_cols = df_student.select_dtypes(include='number').columns

Q1 = df_student[num_cols].quantile(0.25)
Q3 = df_student[num_cols].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

outliers = ((df_student[num_cols] < lower_limit) |
            (df_student[num_cols] > upper_limit))

outliers.sum()
```

```python
df_student[num_cols] = df_student[num_cols].clip(
    lower=lower_limit,
    upper=upper_limit,
    axis=1
)
```

```python
# After outlier handling

plt.figure(figsize=(12, 8))

for i, col in enumerate(num_cols, 1):
    plt.subplot(3, 3, i)
    plt.boxplot(df_student[col])
    plt.title(col.replace("_", " ").title())

plt.tight_layout()
plt.show()
```
